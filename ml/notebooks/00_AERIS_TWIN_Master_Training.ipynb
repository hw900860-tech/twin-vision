## 1. Environment Setup & Shared Data Generator

```python
import os
import math
import random
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, f1_score

print("AERIS-TWIN Master Training Environment Initialized.")
```

## 2. Generate 100,000 Sample Physics-Informed Dataset

```python
random.seed(42)
np.random.seed(42)

# Create data directory
os.makedirs("ml/data", exist_ok=True)
os.makedirs("ml/artifacts", exist_ok=True)

# Generate dataset logic
samples = []
    for i in range(100000):
    alt = random.uniform(0, 30000)
    thr = random.uniform(0, 100)
    amb = random.uniform(-30, 50)
    rho = math.exp(-alt / 27000.0)
    rpm = 2400.0 + (thr / 100.0) * 1600.0 * (0.86 + 0.14 * rho) + np.random.normal(0, 8.0)
    map_kpa = (18.0 + (thr / 100.0) * 14.0 * rho) + np.random.normal(0, 0.4)
    cht_base = 96.0 + (thr / 100.0) * 96.0 + amb * 0.72 - rho * 12.0
    samples.append({
        'altitude': alt, 'throttle': thr, 'ambient_temp': amb, 'air_density_ratio': rho,
        'obs_rpm': rpm, 'obs_map': map_kpa, 'cht_base': cht_base
    })
df = pd.DataFrame(samples)
print(f"Generated dataset shape: {df.shape}")
df.head()
```
